```python
df = pd.read_csv("../../course_assignments/homework9/solar_system.csv")

print("original shape:", df.shape)

df = df.set_index("Attribute")
print("after set_index('Attribute'):", df.shape)

df = df.T
print("after transpose:", df.shape)

df.index.name = "Planet"
print("after naming index:", df.shape)

df.reset_index(inplace = True)
print("after reset_index:", df.shape)

df.columns.name = None
print("after removing columns name:", df.shape)
```

df.set_index("Attribute") makes the Attribute column become the index of the DataFrame. The shape changes from (20,11) to (20,10) because "Attribute" is no longer a regular column.

df.T transposes the DataFrame, which means that the rows and columns switch. Then, each row represents a planet instead of an attribute. 

df.index.name = "Planet" gives the index the name "Planet". This doesn't change the shape.

df.reset_index(inplace=True) turns the planet index into a regular column called "Planet". This changes the shape from (10,20) to (10,21)

df.columns.name = None removes the extra column index name from the DataFrame. This makes the table cleaner and doesn't change the shape.

The new shape is not (11,20) because the "Attribute" column was used as the index before transposing. After transposing, the planet names are stored in the index, not as a normal column. Then reset_index() adds the planet names back as a new column called "Planet", so the final DataFrame has 21 columns.

```python
def attach_units(old_col, new_col, unit):
    new_values = []
    
    for val in df[old_col]:
        new_values.append(val * unit)
    
    df[old_col] = new_values
    df.rename(columns={old_col: new_col}, inplace=True)
```

```python
attach_units("Mass (10^24kg)", "Mass (kg)", 1e24 * u.kg)
```

```python
semi_major_axis = []

for i in range(len(df)):
    value = (df["Perihelion (10^6 km)"][i] + df["Aphelion (10^6 km)"][i]) / 2
    semi_major_axis.append(value * 1e6 * u.km)
```

```python
aphelion_index = df.columns.get_loc("Aphelion (10^6 km)")

df.insert(aphelion_index + 1, "semi-major axis (km)", semi_major_axis)
```

```python
print(df["semi-major axis (km)"])
```

```python
orbital_period_years = []

for val in df["Orbital Period (days)"]:
    orbital_period_years.append((val * u.day).to(u.year))
    
df["Orbital Period (days)"] = orbital_period_years
df.rename(columns={"Orbital Period (days)": "Orbital Period (years)"}, inplace=True)
```

```python
print(df.columns.tolist())
```

```python
print(df["Orbital Period (years)"])
```

```python
earth_period = df.loc[df["Planet"] == "Earth", "Orbital Period (years)"].iloc[0]
print(f"Earth's orbital period is {earth_period.value:.4f} years")
```

Define AU: the average distance between Earth and the Sun.

```python
print(const.au)
```

```python
print(const.au.to(u.km))
```

```python
df["Diameter (AU)"] = [(val * u.km).to(u.AU) for val in df["Diameter (km)"]]

df["Distance from Sun (AU)"] = [(val * 1e6 * u.km).to(u.AU) for val in df["Distance from Sun (10^6 km)"]]

df["Perihelion (AU)"] = [(val * 1e6 * u.km).to(u.AU) for val in df["Perihelion (10^6 km)"]]

df["Aphelion (AU)"] = [(val * 1e6 * u.km).to(u.AU) for val in df["Aphelion (10^6 km)"]]

df["semi-major axis (AU)"] = [val.to(u.AU) for val in df["semi-major axis (km)"]]

df = df.drop(columns=[
    "Diameter (km)",
    "Distance from Sun (10^6 km)",
    "Perihelion (10^6 km)",
    "Aphelion (10^6 km)",
    "semi-major axis (km)"
])
```
