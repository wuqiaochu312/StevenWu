```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("../../course_assignments/homework7/star_data.txt",
skiprows=1)
```

```python
print(data.shape)
```

```python
magnitude = data[:,0]
temperature = data[:,1]
```

```python
print("minimum magnitude:", np.min(magnitude))
print("maximum magnitude:", np.max(magnitude))
```

```python
print("minimum temperature:", np.min(temperature))
print("maximum temperature:", np.max(temperature))
```

```python
brightest_index = np.argmin(magnitude)
dimmest_index = np.argmax(magnitude)
coldest_index = np.argmin(temperature)
hottest_index = np.argmax(temperature)

print("brightest star index:", brightest_index)
print("dimmest star index:", dimmest_index)
print("coldest star index:", coldest_index)
print("hottest star index:", hottest_index)
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

plt.figure(figsize=(8,6))
plt.scatter(log_temperature, log_magnitude, label="Stars")

plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.scatter(log_temperature[brightest_index],log_magnitude[brightest_index],color="black",marker="*",s=300,label="brightest")
plt.scatter(log_temperature[dimmest_index],log_magnitude[dimmest_index],color="purple",marker="*",s=300,label="dimmest")
plt.scatter(log_temperature[coldest_index],log_magnitude[coldest_index],color="orange",marker="*",s=300,label="coldest")
plt.scatter(log_temperature[hottest_index],log_magnitude[hottest_index],color="green",marker="*",s=300,label="hottest")

plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.legend()

plt.show()
```
